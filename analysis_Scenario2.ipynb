## Numの比較

```python
import matplotlib.pyplot as plt
import numpy as np

from collections import defaultdict

import csv

created_prams = []

research_params = [ 0, 50, 99 ]

cpNum = [0] * 10
cpNum_range = [0] * 10
cpNum_dir = [1] * 10
temp = cpNum + cpNum_range + cpNum_dir
resultDirectoryName = ",".join(map(str, temp))


# cpNum1 ~ 10 の時
for idx in range(10):
    cpNum = [0] * 10
    cpNum_range = [0] * 10
    cpNum_dir = [1] * 10
    temp = cpNum + cpNum_range + cpNum_dir

    fig = plt.figure(figsize=(20, 10))
    plt.suptitle(f"checkPoint{idx + 1}", fontsize=16)

    plt.subplots_adjust(hspace=0.4, wspace=0.4)
    detected_bugs_all = defaultdict(int)

    for Range in range(3):
        for Dir in range(1, 5):
            detected_bugs = []
            for Num in range(3):
                cpNum[idx] = research_params[Num]
                cpNum_range[idx] = research_params[Range]
                cpNum_dir[idx] = Dir
                resultDirectoryName = ",".join(map(str, cpNum + cpNum_range + cpNum_dir))

                with open(f"Scenario2/{resultDirectoryName}/detected_bugs.csv", "r", encoding="utf-8") as csvfile:
                    reader = csv.reader(csvfile)
                    bug_num = 0
                    for row in reader:
                        if row != ["null"] and row != ["timeout"]:
                            bug_num += 1
                        detected_bugs_all[row[0]] += 1

                    detected_bugs.append(bug_num)
            ax = fig.add_subplot(3, 4, (3 - Range - 1) * 4 + Dir)
            ax.plot([0, 50, 99], detected_bugs, marker='o', linestyle='-')

            plt.xlabel(f"cpNum")
            plt.xticks([0, 50, 99])
            
            plt.ylabel("Detected Bugs")
            plt.yticks([0, 1, 2, 3, 4, 5])
    fig.tight_layout()
    plt.savefig(f"result_images/many_graphs/Scenario2/Num/checkpoint{idx + 1}.png")
    plt.show()
```
