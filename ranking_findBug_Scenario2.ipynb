# 各パラメータを固定し，その時の不具合検出数を比較する

## 全部可変

```python

from collections import defaultdict

import csv

created_prams = []

research_params = [ 0, 50, 99 ]

cpNum = [0] * 10
cpNum_range = [0] * 10
cpNum_dir = [1] * 10
temp = cpNum + cpNum_range + cpNum_dir
resultDirectoryName = ",".join(map(str, temp))

for idx in range(10):
    # バグ検出数のランキング
    detected_bugs = defaultdict(list)
    print("checkpoint : ", idx)
    
    for Num in range(3):
        cpNum = [0] * 10
        cpNum_range = [0] * 10
        cpNum_dir = [1] * 10
        
        for Range in range(3):
            for Dir in range(1, 5):
                cpNum[idx] = research_params[Num]
                cpNum_range[idx] = research_params[Range]
                cpNum_dir[idx] = Dir
                resultDirectoryName = ",".join(map(str, cpNum + cpNum_range + cpNum_dir))
                with open(f"Scenario2/{resultDirectoryName}/detected_bugs.csv", "r") as csvfile:
                    reader = csv.reader(csvfile)
                    bug_num = 0
                    for row in reader:
                        if row != ["null"] and row != ["timeout"]:
                            bug_num += 1
                    detected_bugs[bug_num].append(f"cpNum={research_params[Num]} cpNum_range={research_params[Range]} cpNum_dir={Dir}")
    print(sorted(detected_bugs.items(), reverse=True))
    print("")
```
